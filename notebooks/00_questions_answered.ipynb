## Big Data 5Vs

**Volume:**  
The project processes hundreds of trailers and thousands of YouTube comments, with a scalable NoSQL-based storage design.

**Velocity:**  
Data is collected via the YouTube API and processed in batches, allowing incremental updates.

**Variety:**  
Structured movie metadata and unstructured text comments are combined, stored and analyzed.

**Veracity:**  
Noisy and incomplete data is handled through validation and multi-stage matching.

**Value:**  
The project provides insight into the relationship between trailer sentiment and movie success.

---

## Four Levels of Data Handling

**Data Source:**  
Movie datasets and the YouTube Data API.

**Data Storage:**  
HBase NoSQL database with separate tables for raw, comment, and joined data.

**Data Analysis:**  
Sentiment analysis.

**Data Output:**  
Visualizations and correlation analysis for interpretation.
