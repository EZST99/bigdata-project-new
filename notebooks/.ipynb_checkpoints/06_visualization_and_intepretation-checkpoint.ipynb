# Visualization and Final Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("../data/final_analysis_dataset.csv")
df.head()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="sentiment_mean",
    y="rating",
    size="comment_count",
    alpha=0.7
)
plt.title("Average Comment Sentiment vs IMDb Rating")
plt.xlabel("Mean Sentiment (VADER)")
plt.ylabel("IMDb Rating")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="comment_count",
    y="revenue",
    alpha=0.7
)
plt.yscale("log")
plt.title("Number of Comments vs Box Office Revenue")
plt.xlabel("Comment Count")
plt.ylabel("Revenue (log scale)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="positive_ratio",
    y="negative_ratio",
    hue="success",
    alpha=0.7
)
plt.title("Positive vs Negative Comment Ratio")
plt.show()
```

```python
corr = df[[
    "sentiment_mean",
    "positive_ratio",
    "negative_ratio",
    "comment_count",
    "rating",
    "revenue",
    "success"
]].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
