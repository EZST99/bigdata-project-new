## Big Data 5Vs

**Volume:**  
The project processes hundreds of trailers and thousands of YouTube comments, with a scalable NoSQL-based storage design.

**Velocity:**  
Data is collected via the YouTube API and processed in batches, allowing incremental updates.

**Variety:**  
Structured movie metadata and unstructured text comments are combined, stored and analyzed.

**Veracity:**  
Noisy and incomplete data is handled through validation and multi-stage matching.

**Value:**  
The project provides insight into the relationship between trailer sentiment and movie success.

---

## Four Levels of Data Handling

**Data Source:**  
Movie datasets and the YouTube Data API.

**Data Storage:**  
HBase NoSQL database with separate tables for raw, comment, and joined data.

**Data Analysis:**  
Sentiment analysis.

**Data Output:**  
Visualizations and correlation analysis for interpretation.

---

### Data Sources
- **YouTube Data API**  
  Used to collect trailer metadata and top-level comments for movie trailers.
- **Kaggle Movie Metadata Dataset (CSV)**  
  Contains IMDb rating, box office revenue and success label for movies.

---

### Tools & Infrastructure
- **Jupyter Notebook** for analysis and documentation
- **HBase** for storing:
  - Trailer data
  - YouTube comments
  - Joined movie–trailer dataset
- **VADER & Apache Spark (PySpark)** for analysis

---

### Access & Configuration
- YouTube API key is required (stored as environment variable `YOUTUBE_API_KEY`)
- HBase connection via HappyBase (`host=hbase`, `port=9090`)
- No usernames or passwords are required beyond local Docker environment
